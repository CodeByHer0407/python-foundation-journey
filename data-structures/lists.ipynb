##### List Comprehension

Basic Syntax            [expression for item in iterable]

with conditional logic    [expression for item in iterable if condition]

Nested List Comprehension [expression for item1 in iterable1 for item2 in iterable2]




```python
### Basic List Comprehension

square=[num**2 for num in range(10)]
print(square)
```

```python
## List comprehension with condition

lst3 = []
for i in range(20):
    if i%2 == 0:
        lst3.append(i)

print(lst3)
```

```python
evens = [x for x in range(20) if x%2 == 0]
print(evens)
```

```python
## Nested List Comphrension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pairs = [[i,j] for i in lst1 for j in lst2]
print(pairs)
```

```python
## List Comprehension with function calls

words = ["hello", "world", "python", "list", "comprehension"]
lengths = [len(word) for word in words]
print(lengths)
```

```python
# Real-world example: filtering active users

users = [
    {"name": "Saloni", "active": True},
    {"name": "Shizi", "active": False},
    {"name": "James", "active": True},
]

def get_active_users(users):
    active_users = [user for user in users if user["active"]]
    return active_users

print(get_active_users(users))
```

### Problem 1: Sum of List Elements

Write a Python program to calculate the sum of all elements in a list.

Approach:
1. Initialize a variable `total` to 0
2. Traverse the list
3. Add each element to `total`
4. Return the final sum

```python
def sum_list(numbers):
    total = 0
    for num in numbers:
        total += num
    return total

n = int(input("Enter the size of the list: "))
my_list = list()
for i in range(n):
    element = int(input("Enter the value to append to the list: "))
    my_list.append(element)

print(f"The sum of the elements of the list: {sum_list(my_list)}")

## or you can use built-in functions

##print(f"Sum: {sum(my_list)}")

```

### Problem 2: Largest Element in a List





Write a Python function that finds and returns the largest element in a given list of integers.

Approach:
1. Initialize first element
2. Traverse the list
3. Compare and update
4. Return result

```python
def find_largest(numbers):
    if not numbers:
        return None
    largest_element = numbers[0]
    for num in numbers[1:]:
        if num > largest_element:
            largest_element = num
    return largest_element

n = int(input("Enter the size of the list: "))
my_list = list()
for i in range(n):
    element = int(input("Enter the value to append to the list: "))
    my_list.append(element)

print(f"The largest element in the list: {find_largest(my_list)}")

## or we can use built-in functions
print(max(my_list))
```

- Time Complexity: O(n)
- Space Complexity: O(1)

### Problem 5: Reverse a List

You are given a list of integers. Write a Python program that reverses the list without using slicing (lst[::-1]). The program should return the reversed list.

Approach:
- Use two pointers — one at the beginning and one at the end of the list.
- Swap elements and move the pointers toward each other until they meet.

```python
def reverse_lst(numbers):
    start = 0
    end = len(numbers) - 1
    while start < end :
        numbers[start], numbers[end] = numbers[end], numbers[start]
        start += 1
        end -= 1
    return numbers 

n = int(input("Enter the size of the list: "))
my_lst = list()
for i in range(n):
    element = int(input("Enter the value to append to the list: "))
    my_lst.append(element)

print(f"List: {my_lst}")
print(f"The reversed list: {reverse_lst(my_lst)}")
```

Time Complexity: O(n)
Space Complexity: O(1)

### Problem 6: Count number of Odd and Even Elements in a List

You are given a list of integers. Write a Python program that counts and returns the number of even and odd numbers in the list.

Approach:

1. Initialize two counters: `even_cnt` and `odd_cnt`
2. Traverse the list
3. Check if each number is divisible by 2
4. Increment the corresponding counter
5. Return the counts of even and odd numbers

```python
def count_even_odd(numbers):
    odd_cnt = 0
    even_cnt = 0
    for num in numbers:
        if num%2 == 0:
            even_cnt += 1
        else:
            odd_cnt += 1
    return even_cnt, odd_cnt

n = int(input("Enter the size of the list: "))
my_lst = list()
for i in range(n):
    element = int(input("Enter the value to append to the list: "))
    my_lst.append(element)

print(f"List: {my_lst}")
even, odd = count_even_odd(my_lst)
print(f"Odd Elements: {odd}, Even Elements: {even}")
```

- Time Complexity: O(n)
- Space Complexity: O(1)

### Problem 7: Maximum difference between two consecutive elements in a list.

Approach:

1. Traverse the list and compute the absolute difference between every pair of consecutive elements. 
2. Maintain a variable to store the maximum difference encountered so far. 
3. Update this value whenever a larger difference is found, and return it after completing the traversal.

```python
def max_consecutive_difference(numbers):
    if len(numbers) < 2 :
        return 0
    max_diff = abs(numbers[0] - numbers[1])
    for i in range(len(numbers) - 1):
        current_diff = abs(numbers[i] - numbers[i+1])
        if current_diff > max_diff:
            max_diff = current_diff
    return max_diff


n = int(input("Enter the size of the list: "))
my_lst = list()
for i in range(n):
    element = int(input("Enter the value to append to the list: "))
    my_lst.append(element)

print(f"List: {my_lst}")
print(f"The maximum consecutive difference in the given list: {max_consecutive_difference(my_lst)}")
```

- Time Complexity: O(n)
- Space Complexity: O(1)

### Problem 8: Merge Two Sorted Lists

You are given two sorted lists of integers. Write a Python function to merge these two sorted lists into one sorted list. The resulting list should also be in non-decreasing order.

Approach:

1. Use two pointers to traverse both sorted lists.
2. Compare the elements at the current positions of both lists.
3. Append the smaller element to the merged list and move that pointer forward.
4. Continue until one of the lists is fully traversed.
5. Append the remaining elements of the other list.
6. Return the merged sorted list.

```python
def merge_two_sorted_lists(lst1, lst2):
    n = len(lst1)
    m = len(lst2)
    left = 0
    right = 0
    lst3 = list()


    while left < n and right < m:
        if lst1[left] < lst2[right]:
            lst3.append(lst1[left])
            left += 1
        else:
            lst3.append(lst2[right])
            right += 1


    while left < n:
        lst3.append(lst1[left])
        left += 1

    while right < m:
        lst3.append(lst2[right])
        right += 1

    return lst3
    

def create_list(x):
    my_lst = list()
    for i in range(x):
        element = int(input("Enter the value to append to the list: "))
        my_lst.append(element)
    return my_lst

n = int(input("Enter the size of List 1: "))
my_lst1 = create_list(n)
print(f"List1: {my_lst1}")

m = int(input("Enter the size of List 2: "))
my_lst2 = create_list(m)
print(f"List2: {my_lst2}")

print(f"The merged sorted list: {merge_two_sorted_lists(my_lst1, my_lst2)}")
```

- Time Complexity: O(n + m)
- Space Complexity: O(n + m)
