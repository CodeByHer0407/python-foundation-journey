# Pattern Problems in Python
Practice problems to strengthen loop logic and nested iteration.

```python
## Rectangular Star Pattern

def pattern1(num):
    for i in range(num):
        for j in range(num):
            print("*", end=" ")
        print()

n = int(input("Enter a number:"))
pattern1(n)

```

```python
## Right-Angled Triangle Star Pattern

def pattern2(num):
    for i in range(num):
        for j in range(i+1):
            print("*", end=" ")
        print()

n = int(input("Enter a number: "))
pattern2(n)
```

```python
## Right-Angled Number Pyramid Pattern


def pattern3(num):
    for i in range(num):
        for j in range(i+1):
            print(j+1, end=" ")
        print()


n = int(input("Enter a number: "))
pattern3(n)
```

```python
## Right-Angled Number Pyramid II Pattern


def pattern4(num):
    for i in range(num):
        for j in range(i+1):
            print(i+1, end=" ")
        print()


n = int(input("Enter a number: "))
pattern4(n)
```

```python
## Inverted Right Pyramid

def pattern5(num):
    for i in range(num):
        for j in range(num-i):
            print("*", end=" ")
        print()

n = int(input("Enter any number: "))
pattern5(n)
```

```python
## Inverted Numbered Right Pyramid

def pattern6(num):
    for i in range(num):
        for j in range(num-i):
            print(f"{j+1}", end=" ")
        print()

n = int(input("Enter any number: "))
pattern6(n)
```

```python
## Increasing Numbered Triangle Pattern

def pattern7(num):
    cnt = 1
    for i in range(num):
        for j in range(i+1):
            print(cnt, end="")
            cnt += 1
        print()
        
n = int(input("Enter any number: "))
pattern7(n)
```

```python
## Star Pyramind

def pattern8(num):
    for i in range(num):
        for j in range(num-i-1):
            print(" ", end=" ")

        for x in range(i*2+1):
            print("*", end=" ")

        for y in range(num-i-1):
            print(" ", end=" ")   
        print()

n = int(input("Enter any number: "))
pattern8(n)


## TC: O(N^2) since nested loops, SC: O(1) as no extra data structures are required.
```

```python
## Inverted Star Pyramid

def pattern9(num):
    for i in range(num):
        for j in range(i):
            print(" ", end=" ")

        for x in range(2*num - (2*i+1)):
            print("*", end=" ")

        for c in range(1, i+1):
            print(" ", end=" ")
        print()

n = int(input("Enter any number: "))
pattern9(n)
```

```python
## Half Diamond Star Pattern

def pattern10(num):
    for i in range(1, 2*num):
        if i <= num :
            for x in range(i):
                print("*", end=" ")
        else:
            for y in range(2*num-i):
                print("*", end=" ")
        print()

n = int(input("Enter any number: "))
pattern10(n)
```

```python
## Hollow Rectangle Pattern


def pattern11(num):
    for i in range(num):
        for j in range(num):
            if i == 0 or j == 0 or i == num-1 or j == num-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

n = int(input("Enter any number: "))
pattern11(n)
```

### ✅ Important pattern trick for DSA

Almost every alphabet pattern uses one of these:

- chr(65 + i)
- chr(64 + i)
- ord('A')

```python
## Increasing Letter Triangle Pattern


def pattern12(num):
    for i in range(num):
        for j in range(i+1):
            print(chr(j+65), end=" ")
        print()

n = int(input("Enter any number: "))
pattern12(n)
```

```python
## Alpha-Ramp Pattern


def pattern13(num):
    for i in range(1, num + 1):
        print(chr(64+i)*i)


n = int(input("Enter any number: "))
pattern13(n)
```

```python
## Reverse Letter Triangle Pattern

def pattern14(num):
    for i in range(num):
        for j in range(num-i):
            print(chr(65+j), end=" ")
        print()


n = int(input("Enter any number: "))
pattern14(n)
```

```python
## Alpha Hill Pattern

def pattern15(num):
    for i in range(num):
        for j in range(num-i-1):
            print(" ", end=" ")
        for x in range(i+1):
            print(chr(65+x), end=" ")
        for y in range(i-1, -1, -1):
            print(chr(65+y), end=" ")

        print()


n = int(input("Enter any number: "))
pattern15(n)
```
