## Inheritance in Python

Inheritance is a fundamental concept in OOPs that allows a class to inherit attributes and methods from another class.

**Inheritance** allows a class (called the **child** or **derived** class) to acquire the attributes and methods of another class (called the **parent** or **base** class).

| Term | Description |
|---|---|
| **Parent / Base class** | The class being inherited from |
| **Child / Derived class** | The class that inherits |
| **`super()`** | Refers to the parent class — used to call its `__init__` or methods |
| **Method Overriding** | Redefining a parent method in the child class |

Inheritance promotes **code reuse** — shared behaviour lives in the parent, while specialised behaviour lives in the child.

## 1. Single Inheritance

A child class inherits from **one** parent class.

````python
## Inheritance (Single Inheritance)

## Parent class

class Car:
    def __init__(self, windows, doors, engine_type):
        self.windows = windows
        self.doors = doors
        self.engine_type = engine_type
    
    def drive(self):
        print(f"The person will drive {self.engine_type} car.")

    
car1 = Car(4, 4, "petrol")
car1.drive()
````

Now we create a `Tesla` class that **inherits** from `Car` and adds its own attribute (`is_selfdrive`) and method (`selfdriving`).

> **`super().__init__(...)`** delegates the initialisation of shared attributes (`windows`, `doors`, `engine_type`) to the parent class, avoiding duplication.

````python
class Tesla(Car):
    def __init__(self, windows, doors, engine_type, is_selfdrive):
        super().__init__(windows, doors, engine_type)
        self.is_selfdrive = is_selfdrive


    def selfdriving(self):
        print(f"Tesla supports self driving: {self.is_selfdrive}")

    

````

````python
tesla1 = Tesla(4, 5, "electric", True)
tesla1.selfdriving()
tesla1.drive()
````

## 2. Multiple Inheritance

A child class inherits from **more than one** parent class, combining their attributes and methods.

```
Animal          Pet
  └──────┬───────┘
         Dog
```

````python
### Multiple Inheritance

## When a class inherits from one more than one base class

## Base class 1
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Subclass must implement this method.")

## Base class 2

class Pet:
    def __init__(self, owner):
        self.owner = owner


## Derived class

class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} says woof!"
````

````python
## create an object 

my_dog = Dog("Joey", "Saloni")
my_dog.speak()
````

````python
print(f"Owner: {my_dog.owner}")
````

## Summary

| Type | Syntax | Description |
|---|---|---|
| Single Inheritance | `class Child(Parent):` | Inherit from one class |
| Multiple Inheritance | `class Child(Parent1, Parent2):` | Inherit from multiple classes |
| `super()` | `super().__init__(...)` | Delegates to the parent constructor |
| Method Overriding | Redefine method in child | Child replaces parent's behaviour |
