## Polymorphism

Polymorphism is a core conecpt in OOPs that allows objects of different classes to be treated as objects of a common superclass. It provides a way to perform a single action in different forms. It is typically achieved through method overriding and interfaces.

**Polymorphism** (Greek: *many forms*) is a core OOP concept that allows objects of different classes to be treated through a shared interface — typically a common parent class or abstract base.

The same function call can produce different behaviour depending on the object it receives.

| Technique | Description |
|---|---|
| **Method Overriding** | A child class provides its own version of a parent method |
| **Polymorphic Functions** | A single function works with objects of different types |
| **Abstract Base Classes** | Force subclasses to implement specific methods |

---

## 1. Method Overriding

A child class provides its own specific implementation of a method already defined in its parent class.  
The parent version is replaced — this is **overriding**.

````python
## Base class

class Animal:
    def speak(self):
        return f"Sound of the animal"


    
## Derived class 1
class Dog(Animal):
    def speak(self):
        return "Woof!"


## Derived class 2
class Cat(Animal):
    def speak(self):
        return "Meow!!"


# Polymorphic function — works with any Animal subclass
def animal_speak(animal):
    print(animal.speak())

my_dog = Dog()
my_cat = Cat()

print(my_dog.speak())
print(my_cat.speak())
animal_speak(my_dog)
````

> **Key idea:** `animal_speak()` doesn't care which subclass it receives — it just calls `.speak()` and each object responds in its own way.

---
## 2. Polymorphism with Functions and Methods

A single function can handle objects of different types as long as they share a common method name.

````python
import math

# Base class
class Shape:
    def area(self):
        return "The area of the figure"  # Generic fallback


# Derived class 1
class Rectangle(Shape):
    def __init__(self, height, width):
        self.height = height
        self.width = width

    def area(self):
        return self.height * self.width


# Derived class 2
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return round(math.pi * self.radius ** 2, 2)  


# Polymorphic function — works with any Shape subclass
def print_area(shape):
    print(f"{shape.__class__.__name__} area: {shape.area()}")


rectangle1 = Rectangle(10, 20)
circle1 = Circle(3)

print_area(rectangle1)  
print_area(circle1)     
````

## 3. Polymorphism with Abstract Base Classes (ABCs)

An **Abstract Base Class (ABC)** defines a required interface — it declares methods that *every* subclass **must** implement.  
If a subclass skips implementing an abstract method, Python raises a `TypeError` at instantiation.

This enforces consistency across all derived classes.

````python
from abc import ABC, abstractmethod

# Abstract base class — cannot be instantiated directly
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        """Every vehicle must define how its engine starts."""
        pass


# Derived class 1
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started."


# Derived class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine started."


# Polymorphic function
def start_vehicle(vehicle):
    print(vehicle.start_engine())


car = Car()
motorcycle = Motorcycle()

start_vehicle(car)         
start_vehicle(motorcycle)  
````

## Summary

| Concept | How it works | Benefit |
|---|---|---|
| **Method Overriding** | Child redefines a parent method | Specialised behaviour per class |
| **Polymorphic Function** | One function, many object types | Flexible, reusable code |
| **Abstract Base Class** | Forces subclasses to implement methods | Consistent interface across classes |
