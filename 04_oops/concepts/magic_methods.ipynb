# Magic Methods in Python

Magic methods (also called **dunder methods** — short for *double underscore*) are special methods Python calls automatically behind the scenes.

You've been using them without realising it:
- `a + b` → Python calls `a.__add__(b)`
- `print(obj)` → Python calls `obj.__str__()`
- `len(obj)` → Python calls `obj.__len__()`

You don't call these methods yourself — Python does. By defining them in your class, you make your objects behave like built-in Python types.

---

### Common magic methods covered here

| Method | Triggered by | Purpose |
|---|---|---|
| `__init__` | `MyClass()` | Initialise a new instance |
| `__str__` | `print(obj)` | User-friendly string display |
| `__repr__` | `repr(obj)` / console | Developer/debug representation |
| `__len__` | `len(obj)` | Return length of object |
| `__getitem__` | `obj[index]` | Get item from container |
| `__setitem__` | `obj[index] = value` | Set item in container |
| `__add__` | `obj1 + obj2` | Addition behaviour |
| `__eq__` | `obj1 == obj2` | Equality comparison |
| `__lt__` | `obj1 < obj2` | Less-than comparison |
| `__enter__` | `with obj as x:` | Enter a context manager |
| `__exit__` | leaving `with` block | Exit a context manager |

## 1. Every class already has magic methods

Even before you define anything, Python gives every class a set of dunder methods by default. You can see them with `dir()`.

```python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def __str__(self):
        return f"{self.name}, {self.age}, {self.gender}"
    
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age}, gender={self.gender})"
```

```python
dir(Person)
```

```python
p1 = Person("Saloni", 26, "Female")
print(p1)
```

```python
print(repr(p1))
```

## 2. `__str__` and `__repr__` — display and debugging

- `__str__` → for end users. Called by `print()`
- `__repr__` → for developers. Should look like valid Python that could recreate the object. Called by `repr()` and shown in the console.

> **Rule of thumb:** `__str__` is what you'd show a user. `__repr__` is what you'd log in an error report.

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"'{self.title}' by {self.author}"
    
    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.pages})"
```

```python
b = Book("Atomic Habits", "James Clear", 320)
print(b)
print(repr(b))
```

## 3. `__len__`, `__getitem__`, `__setitem__` — container behaviour

These three methods let your custom class behave like a Python list or dictionary — supporting `len()`, index access `[]`, and index assignment `[] =`.

```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.__songs = []

    def add(self, song):
        self.__songs.append(song)

    def __len__(self):
        return len(self.__songs)
    
    def __getitem__(self, index):
        return self.__songs[index]
    
    def __setitem__(self, key, value):
        self.__songs[key] = value
```

```python
p = Playlist("Chill Mix")
p.add("Sunflower")
p.add("BlindiNG Lights")
p.add("Stay")
print(len(p))
print(p[1])

p[1] = "Starboy"

for song in p:
    print(song)
```

## 4. `__add__`, `__eq__`, `__lt__` — arithmetic and comparisons

These let you use standard operators (`+`, `==`, `<`) on your own objects, just like Python's built-in types support them.

```python
class Cart:
    def __init__(self, items, total):
        self.items = items
        self.total = total

    def __add__(self, other):       # cart1 + cart2
        combined_items = self.items + other.items
        combined_total = self.total + other.total
        return Cart(combined_items, combined_total)

    def __str__(self):
        return f"Cart({self.items},  ₹{self.total})"
    
    def __lt__(self, other):
        return self.total < other.total
    
    def __eq__(self, other):
        return self.total == other.total
    
cart1 = Cart(["shoes", "shirt"], 1500)
cart2 = Cart(["book", "pen"], 300)

print(cart1 + cart2)
print(cart1 == cart2)
print(cart2 < cart1)
```

## 5. `__enter__` and `__exit__` — the `with` statement

Ever wondered how `with open('file.txt')` automatically closes the file even if an error occurs? That's these two methods — they define a **context manager**.

- `__enter__` → runs when you enter the `with` block, returns the resource
- `__exit__` → runs when you leave the `with` block, even if an exception was raised

```python
class DatabaseConnection:
    def __init__(self, db_name):
        self.db_name = db_name

    def __enter__(self):                
        print(f"Connecting to {self.db_name}...")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):   
        print(f"Closing connection to {self.db_name}.")
        return False                    

    def query(self, sql):
        print(f"Running: {sql}")


with DatabaseConnection("users_db") as db:
    db.query("SELECT * FROM users")

```
