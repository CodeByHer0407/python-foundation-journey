## Classes and Objects


Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modeling real-world scenarios using classes and objects. 

| Concept | Description |
|---|---|
| **Class** | A blueprint for creating objects |
| **Object** | An instance of a class |
| **Attribute** | A variable that belongs to an object |
| **Method** | A function that belongs to an object |

OOP helps model real-world scenarios in code — making programs easier to read, reuse, and maintain.


## 1. Defining a Class and Creating Objects

A **class** is defined using the `class` keyword. Once defined, you can create multiple **objects** (instances) from it.

```python
### A class is a blueprint for creating objects, attributes, methods.

class Car:
    pass


my_audi = Car()
my_benz = Car()

print(type(my_audi))
print(type(my_benz))
```

> **Note:** Both `my_audi` and `my_benz` are separate objects of type `Car`, each occupying a different memory location.

```python
print(my_audi)
```

## 2. The Constructor (`__init__`) and Instance Variables

The `__init__` method is the **constructor** — it runs automatically when a new object is created.  
It initialises **instance variables** (attributes unique to each object).

```python
### Instance Variable and Methods

class Dog:

    ## Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age


## create objects
my_pet = Dog("Phoebe", 3)
print(my_pet)
print(my_pet.name)
print(my_pet.age)
```

```python
my_pet2 = Dog("Lucy", 4)
print(my_pet2)
print(my_pet2.name)
print(my_pet2.age)
```

## 3. Instance Methods

**Instance methods** are functions defined inside a class that act on the object's own data.  
They always take `self` as their first parameter, giving them access to instance variables.

```python
## Define a class with instance method

class Dog:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof!!")


dog1 = Dog("buddy", 3)
dog1.bark()
dog2 = Dog("Lucy", 4)
dog2.bark()
```

## 4. Real-World Modeling: Bank Account

OOP shines when modelling real-world entities. Here we model a `BankAccount` with:

- `__init__` — sets the account holder name and starting balance
- `__str__` — returns a human-readable description of the account
- `deposit()` — adds funds to the balance
- `withdraw()` — deducts funds with an insufficient balance guard
- `check_balance()` — returns the current balance

```python
## Modeling a Bank Account

## Define a class for bank account

class BankAccount:

    def __init__(self, account_holder, balance=0):
        """
        Initialise account with the holder's name and an optional starting balance.
        Default balance is 0 if not provided.
        """
        self.account_holder = account_holder
        self.balance = balance

    def __str__(self):
        """Human-readable string representation of the account."""
        return f"Account Holder: {self.account_holder} | Balance: {self.balance}"

    def deposit(self, amount):
        """Adds the given amount to the balance."""
        self.balance += amount
        print(f"{amount} deposited. New balance: {self.balance}")

    def withdraw(self, amount):
        """Withdraws amount if sufficient funds exist, otherwise shows a warning."""
        if self.balance < amount:          # Guard: reject if funds are insufficient
            print("Insufficient funds! Withdrawal cancelled.")
        else:
            self.balance -= amount
            print(f"{amount} withdrawn. Remaining balance: {self.balance}")

    def check_balance(self):
        """Returns the current account balance."""
        return self.balance


# --- Demo ---
my_account = BankAccount("Saloni", 1500)
print(my_account)                               # Uses __str__

my_account.deposit(500)
print("Balance:", my_account.check_balance())   # 2000

my_account.withdraw(1000)
print("Balance:", my_account.check_balance())   # 1000

my_account.withdraw(5000)                       # Triggers insufficient funds warning
    

```

## Summary

| Feature | Syntax | Purpose |
|---|---|---|
| Define a class | `class ClassName:` | Create a blueprint |
| Constructor | `def __init__(self, ...)` | Initialise object attributes |
| Instance variable | `self.variable = value` | Store data on the object |
| Instance method | `def method(self):` | Define object behaviour |
| String representation | `def __str__(self):` | Human-readable object output |
| Create an object | `obj = ClassName(args)` | Instantiate the class |
| Access attribute | `obj.attribute` | Read object data |
| Call a method | `obj.method()` | Trigger object behaviour |
