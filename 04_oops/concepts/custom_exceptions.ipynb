## Custom Exceptions (Raise and Throw an exception)

```python
class ExamEligibilityError(Exception):
    pass

class DOBException(ExamEligibilityError):
    def __init__(self, age):
        self.age = age
        super().__init__(f"Age {age} is not eligible. Must be between 20 and 30.")
```

```python
from datetime import datetime
year = int(input("Please enter your DOB: "))
current_age = datetime.now().year - year

try:
    if current_age <= 30 and current_age >= 20:
        print(f"Go ahead, you are eligible for the exam!")
    else:
        raise DOBException
except DOBException:
    print("Age is outside the eligible range of 20-30.")
```
