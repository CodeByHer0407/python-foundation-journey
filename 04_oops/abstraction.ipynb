# Abstraction in Python

**Abstraction** is the concept of hiding complex implementation details and exposing only what is necessary. It answers the question: *"What does this do?"* — not *"How does it do it?"*

### Why does it matter?
- Reduces complexity for the caller
- Enforces a consistent interface across multiple classes
- Makes code easier to swap, extend, and maintain

### How Python implements abstraction
Python uses **Abstract Base Classes (ABC)** from the `abc` module. An abstract class:
- Acts as a **blueprint** — it cannot be instantiated directly
- Defines **abstract methods** that every subclass *must* implement
- Can also include **concrete methods** shared across all subclasses

---

```python
from abc import ABC, abstractmethod

# Abstract base class 
class Vehicle(ABC):

    @abstractmethod
    def drive(self):
        """Every vehicle must implement how it drives."""
        pass

    @abstractmethod
    def start_engine(self):
        """Every vehicle must implement how its engine starts."""
        pass

    def fuel_up(self):
        print(f"{self.__class__.__name__}: fuelling up...")


class Car(Vehicle):
    def start_engine(self):
        print("Car: turning ignition key, cylinders firing...")

    def drive(self):
        print("Car: engaging gears, moving forward.")


class Bike(Vehicle):
    def start_engine(self):
        print("Bike: kick-starting engine...")

    def drive(self):
        print("Bike: balancing and accelerating.")



def operate_vehicle(vehicle: Vehicle):
    vehicle.start_engine()
    vehicle.drive()
    vehicle.fuel_up()   


car = Car()
operate_vehicle(car)

print()

my_bike = Bike()
operate_vehicle(my_bike)
```

### What happens if you try to instantiate the abstract class directly?

This is a key behaviour to understand — Python enforces the contract at runtime.

```python
try:
    v = Vehicle()  # attempting to instantiate an abstract class
except TypeError as e:
    print(f"TypeError: {e}")

# Output: Can't instantiate abstract class Vehicle
# with abstract methods drive, start_engine
```

### What if a subclass forgets to implement an abstract method?

```python
class BrokenCar(Vehicle):
    def start_engine(self):
        print("Engine started")
    # drive() is NOT implemented — Python will catch this

try:
    b = BrokenCar()
except TypeError as e:
    print(f"TypeError: {e}")

# Python forces you to complete the contract — this is the whole point of abstract methods
```

```python
from abc import ABC, abstractmethod

# ABSTRACTION 
class CoffeeMachine(ABC):

    @abstractmethod
    def make_coffee(self):
        pass

    @abstractmethod
    def clean(self):
        pass


# ENCAPSULATION + ABSTRACTION together
class Nespresso(CoffeeMachine):

    def __init__(self):
        self.__water_level = 500      
        self.__capsules = 10          
        self.__is_on = False          

    # controlled way to READ private data
    def get_status(self):
        print(f"Water: {self.__water_level}ml | Capsules: {self.__capsules}")

    def turn_on(self):
        self.__is_on = True
        print("Nespresso: machine on.")

    # abstract method fulfilled — complex logic hidden inside
    def make_coffee(self):
        if not self.__is_on:
            print("Turn the machine on first!")
        elif self.__water_level < 50:
            print("Refill water!")
        elif self.__capsules == 0:
            print("No capsules left!")
        else:
            self.__water_level -= 50   
            self.__capsules -= 1       
            print("Nespresso: piercing capsule, brewing at 19 bar... ☕")

    # abstract method fulfilled
    def clean(self):
        print("Nespresso: running descale cycle.")


machine = Nespresso()
machine.turn_on()
machine.get_status()    
machine.make_coffee()
machine.make_coffee()
machine.get_status()    
```

---
## Summary

| Concept | What it does | Python tool |
|---|---|---|
| Abstract class | Blueprint — cannot be instantiated | `class MyClass(ABC)` |
| Abstract method | Forces subclasses to implement the method | `@abstractmethod` |
| Concrete method | Shared logic defined once in the base class | Regular `def` in the ABC |

### Key takeaways
- An abstract class defines **what** must exist, not **how**
- Python raises `TypeError` at instantiation if any abstract method is missing
- The calling code only depends on the **interface** (`Vehicle`, `PaymentGateway`, `Shape`) — making it trivial to add new implementations without touching existing code
- Combine with **encapsulation** (private attributes) for fully robust OOP design

> **Rule of thumb:** If you find yourself writing the same `if isinstance(x, Car)` checks repeatedly, that's a sign you need abstraction.
