# Recursion — Introduction

> *Recursion is when the solution of a problem depends on the same smaller version of itself.*

---

---
## 1. The Problem with Separate Functions

Imagine you want to print numbers from 5 down to 1.

A naive approach: write a **separate function for each step**.
This works — but notice how we're doing **almost identical work** in every function.
That's a code smell. 🚩

````python
# ❌ Naive approach — repetitive, doesn't scale

def printNumFive(num):
    print(num)
    printNumFour(num - 1)

def printNumFour(num):
    print(num)
    printNumThree(num - 1)

def printNumThree(num):
    print(num)
    printNumTwo(num - 1)

def printNumTwo(num):
    print(num)
    printNumOne(num - 1)

def printNumOne(num):
    print(num)   # no further call — chain stops here

printNumFive(5)
````

---
## 2. Key Insight — Functions Stay in Memory

From the exercise above, two important things become clear:

| # | Observation |
|---|---|
| 1 | **Functions wait in memory** until they are fully resolved |
| 2 | **Only after a function finishes** does it get removed from the call stack |

This is what makes recursion possible — each call **waits patiently** in the stack while the next one runs.

### What is the Call Stack?
Think of it like a stack of plates 🍽️ — last in, first out (LIFO).
```
printNumFive(5)  ← called first, resolves last
  printNumFour(4)
    printNumThree(3)
      printNumTwo(2)
        printNumOne(1)  ← called last, resolves first
```

---
## 3. Function Calling Itself

Since every function above was doing the **same thing**, we can collapse them into one:
- Print the number
- Call itself with `num - 1`
- Stop when `num == 1` ← this is called the **base case**

> ⚠️ **The base case is critical.** Without it, the function calls itself forever → `RecursionError: maximum recursion depth exceeded`.

````python
# ✅ Recursive approach — clean, scalable

def printNum(num):
    print(num)
    if num == 1:       # base case — stop the recursion
        return
    printNum(num - 1)  # recursive call — same function, smaller problem

printNum(5)
````

---
## 4. How the Call Stack Works

Let's trace exactly what happens when `printNum(5)` is called:

### 📥 Winding up (calls being pushed onto the stack)
```
printNum(5) → prints 5, calls printNum(4)
  printNum(4) → prints 4, calls printNum(3)
    printNum(3) → prints 3, calls printNum(2)
      printNum(2) → prints 2, calls printNum(1)
        printNum(1) → prints 1, hits base case → returns ✅
```

### 📤 Unwinding (calls being popped off the stack)
```
        printNum(1) ← returns, removed from stack
      printNum(2)   ← resumes, returns, removed
    printNum(3)     ← resumes, returns, removed
  printNum(4)       ← resumes, returns, removed
printNum(5)         ← resumes, returns, removed ✅
```

Output: `5 4 3 2 1` — each number printed during the **winding** phase.

````python
# What happens without a base case? → RecursionError
import sys

def infinite_recursion(n):
    print(n)
    infinite_recursion(n - 1)   # no base case!

try:
    sys.setrecursionlimit(20)   # keeping it small for demo
    infinite_recursion(5)
except RecursionError as e:
    print(f"\nRecursionError: {e}")
finally:
    sys.setrecursionlimit(1000)  # reset to default
````

## 5. Revision Summary

| Concept | Key Takeaway |
|---|---|
| **Recursion** | A function that calls itself to solve a smaller version of the same problem |
| **Base case** | The condition that stops the recursion — always required |
| **Call stack** | Functions pile up (LIFO) and resolve in reverse order |
| **Winding** | Phase where recursive calls are being made and pushed to stack |
| **Unwinding** | Phase where calls finish and are popped off the stack |
| **RecursionError** | Happens when there's no base case — stack overflows |

### The Recursion Template
```python
def recursive_function(problem):
    if base_case:          # ← when to stop
        return
    # do some work
    recursive_function(smaller_problem)   # ← same problem, smaller input
```

### Mental model
```
5 → 4 → 3 → 2 → 1 (base case)
         ↑
each step is the same problem with a smaller input
```
