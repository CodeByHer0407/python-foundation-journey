## PMI — Principle of Mathematical Induction

> Recursion is **based on PMI**. Understanding PMI = mastering recursion.

PMI is used to **prove that a statement `f(n)` is true for all n**.

### 3 Steps of PMI (= 3 Steps of Recursion)

| Step | PMI | Recursion Equivalent |
|------|-----|----------------------|
| **1** | Prove `F(0)` or `F(1)` is true | **Base Case** |
| **2** | *Assume* `F(k)` is true (Induction Hypothesis) | **Assume `f(n-1)` works** |
| **3** | Using step 2, prove `F(k+1)` is true | **Solve for `f(n)` using `f(n-1)`** |

> 💡 **Key insight:** We don't need to trace through every function call. We just need to know the **base** and how to take **one step forward** — the rest follows automatically!

---

## Example 1: Factorial using Recursion

### Mathematical Relation:
$$n! = n \times (n-1)!$$
$$\text{fact}(n) = n \times \text{fact}(n-1)$$

### PMI Breakdown:
- **Base case:** `fact(0) = 1` or `fact(1) = 1`
- **Assume:** `fact(n-1)` works correctly
- **Prove:** `fact(n) = n * fact(n-1)` ✅

````python
def factorial(n):
    """Returns factorial of n using recursion."""
    # Base Case
    if n == 0 or n == 1:
        return 1
    # Recursive Case: fact(n) = n * fact(n-1)
    return n * factorial(n - 1)

# Test
for i in range(6):
    print(f"factorial({i}) = {factorial(i)}")
````

### 📊 Call Stack Trace for `factorial(4)`:
```
factorial(4)
  └─ 4 * factorial(3)
           └─ 3 * factorial(2)
                    └─ 2 * factorial(1)
                               └─ returns 1   ← BASE CASE
                    └─ returns 2
           └─ returns 6
  └─ returns 24
```

## Example 2: Sum of N Natural Numbers

### Formula (proved via PMI):
$$\sum_{i=1}^{n} i = \frac{n(n+1)}{2}$$

### PMI Proof (from notes):

**Step 1 — Base Case:**
- `F(0)`: LHS = Σ0 = 0, RHS = 0(0+1)/2 = 0 ✅  
- `F(1)`: LHS = Σ1 = 1, RHS = 1(1+1)/2 = 1 ✅

**Step 2 — Assume `F(k)` is true:**
$$\Sigma k = \frac{k(k+1)}{2}$$

**Step 3 — Prove `F(k+1)` is true:**
$$\Sigma(k+1) = (k+1) + \Sigma k = (k+1) + \frac{k(k+1)}{2} = \frac{(k+1)(k+2)}{2} \checkmark$$

### Recursive Relation:
$$\text{sum}(n) = n + \text{sum}(n-1)$$

````python
def sum_n(n):
    """Returns sum of first n natural numbers using recursion."""
    # Base Case
    if n == 0:
        return 0
    # Recursive Case: sum(n) = n + sum(n-1)
    return n + sum_n(n - 1)

# Test
for i in range(6):
    expected = i * (i + 1) // 2
    result = sum_n(i)
    print(f"sum_n({i}) = {result}  |  formula = {expected}  |  match: {result == expected}")
````

### 📊 Call Stack Trace for `sum_n(4)`:
```
sum_n(4)
  └─ 4 + sum_n(3)
           └─ 3 + sum_n(2)
                    └─ 2 + sum_n(1)
                               └─ 1 + sum_n(0)
                                          └─ returns 0  ← BASE CASE
                               └─ returns 1
                    └─ returns 3
           └─ returns 6
  └─ returns 10
```

## 💪 Example 3: Power of 2 (2ⁿ)

### Recursive Relation:
$$2^n = 2 \times 2^{n-1}$$
$$\text{pow}(2, n) = 2 \times \text{pow}(2, n-1)$$

````python
def power_of_2(n):
    """Returns 2^n using recursion."""
    # Base Case
    if n == 0:
        return 1
    # Recursive Case: 2^n = 2 * 2^(n-1)
    return 2 * power_of_2(n - 1)

# Test
for i in range(8):
    print(f"2^{i} = {power_of_2(i)}")
````

## 🧩 The PMI Framework for ANY Recursion Problem

Apply this 3-step framework to every recursion question:

```
┌─────────────────────────────────────────────────────────┐
│  STEP 1: BASE CASE                                      │
│  → Identify the smallest version of the problem         │
│  → f(0) = ? or f(1) = ?                                 │
├─────────────────────────────────────────────────────────┤
│  STEP 2: ASSUME f(n-1) WORKS                            │
│  → Trust that the smaller version gives correct answer  │
│  → (Induction Hypothesis — don't question it!)          │
├─────────────────────────────────────────────────────────┤
│  STEP 3: SOLVE f(n) USING f(n-1)                        │
│  → Express the answer for n in terms of (n-1)           │
│  → This becomes your return statement                   │
└─────────────────────────────────────────────────────────┘
```

> ✨ **You don't need to trace all function calls in your head. Just know the base and one step forward — recursion handles the rest!**

## ⚡ Common Pitfalls

1. **Missing base case** → `RecursionError: maximum recursion depth exceeded`
2. **Base case never reached** → infinite recursion (e.g., decrementing wrong variable)
3. **Wrong return** → forgetting to `return` the recursive call
