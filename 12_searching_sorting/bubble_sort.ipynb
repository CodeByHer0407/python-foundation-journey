# Bubble Sort

## What is Bubble Sort?
Bubble Sort repeatedly compares **adjacent elements** and swaps them if they are in the wrong order.
After each full pass, the **largest unsorted element "bubbles up"** to its correct position at the end.

## Key Idea
- We do `n-1` passes over the list.
- After each pass, the last `i` elements are already sorted — so we don't need to check them again.
- This is why the inner loop **shrinks** by `i` each time.

## Time Complexity
| Case | Complexity |
|------|------------|
| Best | O(n²) |
| Worst | O(n²) |
| Space | O(1) — sorts in-place |

> **Note:** This is the basic version. An optimized version can exit early if no swaps happen in a pass (means already sorted).

## Helper — Create List

````python
def create_list(n):
    """Takes n inputs from user and returns a list."""
    my_lst = []
    for i in range(n):
        ele = int(input(f"Enter element {i+1}: "))
        my_lst.append(ele)
    return my_lst
    
````

## Helper — Swap

Swaps two elements **in-place** using their **index positions** (not values).

```
swap(lst, 0, 1)  →  swaps lst[0] and lst[1]
```

> ⚠️ Common mistake: passing values instead of indices. Always pass `j` and `j+1`, not `lst[j]` and `lst[j+1]`.

````python
def swap(lst, val1, val2):
    """Swaps elements at index val1 and val2 in lst (in-place)."""
    lst[val1], lst[val2] = lst[val2], lst[val1]
    return lst


````

## Implementation — Iterative

### How the loops work:

```
lst = [5, 6, 1, 3]   →   size = 4

Outer loop: i = 0, 1, 2  (3 passes = size-1)
Inner loop: j = 0 to (size-1-i)  shrinks each pass
```

### Pass-by-pass trace for [5, 6, 1, 3]:
```
Pass 1 (i=0):  [5,6,1,3] → [5,1,3,6]   ← 6 bubbles to end
Pass 2 (i=1):  [5,1,3,6] → [1,3,5,6]   ← 5 bubbles to position
Pass 3 (i=2):  [1,3,5,6] → [1,3,5,6]   ← already sorted, no swaps
```

### Why `size-1-i` in the inner loop?
- After pass `i`, the last `i` elements are sorted and frozen.
- No need to compare them again → saves unnecessary work.

````python
def bubble_sort(lst1):
    size = len(lst1)
    for passes in range(size-1):        # we need (n-1) passes
        for j in range(size-1-passes):  # inner loop shrinks each pass
            if lst1[j] > lst1[j+1]:     # if left > right → out of order
                swap(lst1, j, j+1)      # swap them
    return lst1                         # sorted in-place, same list object
````

````python
n = int(input("Enter the size of the list: "))
lst1 = create_list(n)
print(f"List: {lst1}")


print(f"Before Sorting: {lst1}")
result = bubble_sort(lst1)
print(f"After Sorting: {result}")
````
