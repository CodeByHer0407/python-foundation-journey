# Binary Search

### What's the idea?
Instead of checking every element one by one,  
we look at the **middle element** and ask —  
is my target smaller or bigger?

- If smaller → search the **left half**
- If bigger  → search the **right half**
- Keep doing this until found (or not)

> ⚠️ Only works on a **sorted** list!

---

### Dry Run

```
lst    = [11, 22, 33, 44, 55]
target = 44

step 1 →  start=0  end=4  mid=2  →  lst[2] = 33  →  33 < 44  →  go RIGHT  →  start = 3
step 2 →  start=3  end=4  mid=3  →  lst[3] = 44  →  found! return 3
```

```
lst    = [11, 22, 33, 44, 55]
target = 21  (not in list)

step 1 →  start=0  end=4  mid=2  →  lst[2] = 33  →  33 > 21  →  go LEFT   →  end = 1
step 2 →  start=0  end=1  mid=0  →  lst[0] = 11  →  11 < 21  →  go RIGHT  →  start = 1
step 3 →  start=1  end=1  mid=1  →  lst[1] = 22  →  22 > 21  →  go LEFT   →  end = 0
step 4 →  start=1  end=0  →  start > end  →  return -1
```

---

## Helper — Create List

````python
def create_list(n):
    my_lst = []
    for i in range(n):
        ele = int(input(f"Enter element {i+1}: "))
        my_lst.append(ele)
    return my_lst
    
````

## Implementation — Iterative


````python
def binary_search(lst1, target):
    start = 0
    end = len(lst1) - 1
    while start <= end:
        middle = (start + end) // 2
        if lst1[middle] == target:
            return middle
        
        elif lst1[middle] < target:
            start = middle + 1          # go right
        elif lst1[middle] > target:
            end = middle - 1            # go left

    return -1       # not found
````

````python
n = int(input("Enter the size of the list: "))
lst1 = create_list(n)
print(f"List: {lst1}")
lst1.sort()  # must be sorted!
print(f"Sorted List: {lst1}")

target_ele = int(input("Enter the element you would like to search for: "))
result = binary_search(lst1, target_ele)
if result == -1 :
    print(f"Sorry, can't find {target_ele} in {lst1}")
else:
    print(f"{target_ele} lies at index {result}")
````

---

### Complexity

| | Best | Average | Worst |
|--|------|---------|-------|
| Time | O(1) | O(log n) | O(log n) |
| Space | O(1) | O(1) | O(1) |

**Why log n?**  
Every step cuts the list in half.  
For 1000 elements → max ~10 steps.  
For 1,000,000 elements → max ~20 steps. That's it!

---

### Real World Uses

- 📖 Dictionary / phonebook — sorted alphabetically, so binary search works perfectly
- 🎮 Guess the number game — always guess the middle, eliminate half
- 🔢 Database indexing — how DBs find rows super fast under the hood
- 📦 Git bisect — finds the bad commit by halving history

---

### Binary vs Linear Search (quick reminder)

| | Linear | Binary |
|--|--------|--------|
| Time | O(n) | O(log n) |
| Needs sorted list? | No | Yes |
| 1M elements | ~1,000,000 steps | ~20 steps |

---

### Things to NOT forget
- List must be **sorted** first
- `while start <= end` — the `=` matters, handles single element case
- `middle = (start + end) // 2` — integer division
- `start = middle + 1` and `end = middle - 1` — avoid infinite loop
