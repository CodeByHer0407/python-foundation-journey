# Insertion Sort

Like sorting playing cards in your hand.
Pick one card at a time and insert it into its correct position among the already sorted cards.

| | |
|---|---|
| Time | O(n²) |
| Space | O(1) — in-place |
| Best case | O(n) — already sorted |

````python
def create_list(n):
    my_lst = []
    for i in range(n):
        ele = int(input(f"Enter element {i+1}: "))
        my_lst.append(ele)
    return my_lst
````

## How it works

```
[5, 6, 1, 3]

Pick 6 → 6 > 5, already in place     → [5, 6, 1, 3]
Pick 1 → shift 6, shift 5, insert 1  → [1, 5, 6, 3]
Pick 3 → shift 6, shift 5, insert 3  → [1, 3, 5, 6] ✅
```

**Key variables:**
- `currentCard` — element being inserted
- `correctPosition` — pointer moving left through sorted region
- Shift elements right until correct spot is found, then place

````python
def insertion_sort(lst1):
    size = len(lst1)
    for current in range(1, size):          # start from index 1 (first element is already 'sorted')
        currentCard = lst1[current]         # element to be inserted
        correctPosition = current - 1      # start comparing from the element just before

        while correctPosition >= 0:
            if lst1[correctPosition] < currentCard:  # found correct spot
                break
            else:
                lst1[correctPosition + 1] = lst1[correctPosition]  # shift right
                correctPosition -= 1
                lst1[correctPosition + 1] = currentCard             # place card
    return lst1
````

````python
n = int(input("Enter the size of the list: "))
lst1 = create_list(n)
print(f"List: {lst1}")
print(f"Before Sorting: {lst1}")    #11, 25, 12, 34, 50, 22


result2 = insertion_sort(lst1)
print(f"After Sorting: {result2}")  
````
