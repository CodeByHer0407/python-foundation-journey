# Selection Sort 

## What is Selection Sort?
Selection Sort works by repeatedly **finding the minimum element** from the unsorted part and placing it at the correct position.

Each pass **selects** the smallest remaining element and puts it in its final place.

## Key Idea
- Divide the list into two parts: **sorted (left)** and **unsorted (right)**.
- Each pass, find the minimum in the unsorted part → swap it to the front of unsorted.
- After `i` passes, the first `i` elements are sorted and frozen.
- Only **one swap per pass** — this is selection sort's main advantage.

## Time & Space Complexity
| Case | Complexity |
|------|------------|
| Best | O(n²) |
| Worst | O(n²) |
| Swaps | O(n) — at most n-1 swaps total |
| Space | O(1) — sorts in-place |

> **Compared to Bubble Sort:** Same O(n²) time, but far fewer swaps. Better when swapping is expensive.

## Helper — Create List

````python
def create_list(n):
    my_lst = []
    for i in range(n):
        ele = int(input(f"Enter element {i+1}: "))
        my_lst.append(ele)
    return my_lst
````

## Helper — Swap

Swaps two elements **in-place** using their **index positions** (not values).

```
swap(lst, 0, 2)  →  swaps lst[0] and lst[2]
```

> ⚠️ Always pass indices, not values.

````python
def swap(lst, val1, val2):
    lst[val1], lst[val2] = lst[val2], lst[val1]
    return lst
````

## Implementation — Selection Sort

### How the loops work:
```
lst = [5, 6, 1, 3]   →   size = 4

Outer loop: i = 0, 1, 2  (size-1 passes)
  → i marks the start of the unsorted region
  → current_min tracks the INDEX of the smallest element found so far

Inner loop: j = i+1 to size
  → scans the unsorted region to find the true minimum INDEX
  → only updates current_min, NO swapping yet

After inner loop: swap lst[i] with lst[current_min]  ← ONE swap per pass
```

### Pass-by-pass trace for [5, 6, 1, 3]:
```
Pass 0 (i=0): scan [6,1,3] → min is 1 at index 2 → swap(lst, 0, 2)
              [1, 6, 5, 3]

Pass 1 (i=1): scan [5,3]   → min is 3 at index 3 → swap(lst, 1, 3)
              [1, 3, 5, 6]

Pass 2 (i=2): scan [6]     → min is 6 at index 3 → swap(lst, 2, 3)... wait 5<6, no swap needed
              [1, 3, 5, 6]  ✅
```

### ⚠️ Common Bug — swapping inside the inner loop


The inner loop's only job is to **find** the minimum — swapping should happen **after**.

````python
def selection_sort(lst1):
    size = len(lst1)
    for i in range(size-1):
        current_min = i
        for j in range(i+1, size):
            if lst1[j] < lst1[current_min]:
                swap(lst1, j, current_min)
    return lst1
````

````python
n = int(input("Enter the size of the list: "))
lst1 = create_list(n)
print(f"List: {lst1}")
print(f"Before Sorting: {lst1}")    #11, 25, 12, 34, 50, 22

#result = bubble_sort(lst1)
result1 = selection_sort(lst1)
print(f"After Sorting: {result1}")  
````
