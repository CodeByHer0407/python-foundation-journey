# Linear Search

## What is Linear Search?
- Linear search sequentially checks each element in a list until the target is found or the list ends.
- It is the simplest searching algorithm and works on **both sorted and unsorted** data.

## Time & Space Complexity

| Case | Time Complexity | When? |
|------|----------------|-------|
| Best | O(1) | Target is the first element |
| Average | O(n/2) → O(n) | Target is somewhere in the middle |
| Worst | O(n) | Target is last or not present |


| Space | Complexity | Reason |
|-------|-----------|--------|
| Iterative | O(1) | No extra memory used |
| Recursive | O(n) | Call stack grows with each recursive call |

---

## Real World Use Cases

| Scenario | Why Linear Search? |
|----------|-------------------|
| 🔍 Searching contacts on phone | List may be unsorted or small |
| 📋 Checking if a username exists in a small DB | Simple scan is sufficient |
| 🎵 Finding a song in a playlist | Unsorted, sequential access |
| 🚨 Scanning logs for an error message | One-time scan of raw log file |

> **When NOT to use:** Large sorted datasets → use Binary Search O(log n) instead.

## Helper — Create List

````python
def create_list(n):
    my_lst = []
    for i in range(n):
        ele = int(input(f"Enter element {i+1}: "))
        my_lst.append(ele)
    return my_lst
    
````

## Implementation — Iterative

Traverse the list one by one and return the index on a match.

````python
def linear_search(lst, target):
    for i in range(len(lst)):
        if lst[i] == target:
            return i
    return -1
````

````python
n = int(input("Enter the size of the list: "))
lst1 = create_list(n)
print(f"List: {lst1}")

target_ele = int(input("Enter the element you would like to search for: "))
result = linear_search(lst1, target_ele)
if result == -1 :
    print(f"Sorry, can't find {target_ele} in {lst1}")
else:
    print(f"{target_ele} lies at index {result}")
````
