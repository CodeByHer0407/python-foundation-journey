## File Operation - Read And Write Files

File handling is a crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary. 

```python
## Read a whole file

with open('example.txt', "r") as file:
    content = file.read()
    print(content)
```

```python
## Read a file line by line

with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())
```

```python
import os

# Show the current working directory of your notebook
print(os.getcwd())

# Show the absolute path of the file you just wrote
print(os.path.abspath("example.txt"))
```

```python
## Writing a file (Overwriting)

with open('example.txt', "w") as file:
    file.write("Hey, I am Saloni.\n")
    file.write("Welcome to my World!!\n")
```

```python
## Writing a file (without overwriting)

with open('example.txt', "a") as file:
    file.write("Append operation taking place now...")
```

```python
## Writing a list of lines to a file

lines = ['First line \n', 'Second line \n', 'Third line\n']
with open('example.txt', "a") as file:
    file.writelines(lines)
```

```python
## Binary Files

# Writing to a binary file

data = b'\x00\x01\x02\x03\x04'
with open('examples.bin', "wb") as file:
    file.write(data)
```

```python
## Reading a binary file

with open("examples.bin", "rb") as file:
    content = file.read()
    print(content)
```

```python
### Read the content from a source text file and write it to a destination text file

with open('source.txt', "r") as source_file:
    content = source_file.read()

with open('destination.txt', "w") as destination_file:
    destination_file.write(content)
```
