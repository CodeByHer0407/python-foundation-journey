# Decorators in Python

---
## 1. What are Decorators?

Decorators are a powerful and elegant feature in Python that allow you to **modify or extend the behavior of a function** without changing its source code.

### Common real-world use cases:
| Use Case | Description |
|---|---|
| 📋 **Logging** | Automatically log function calls and results |
| 🔐 **Authentication** | Check permissions before running a function |
| ⏱️ **Performance Measurement** | Measure and report execution time |
| 🔁 **Retry Logic** | Automatically retry failed operations |

### Prerequisites
To understand decorators, you first need to understand two concepts:
- **Functions as first-class objects** — functions can be assigned, passed, and deleted like variables
- **Closures** — inner functions that capture variables from their enclosing scope

---
## 2. Functions as First-Class Objects

In Python, functions are **first-class objects** — they can be:
- Assigned to variables
- Passed as arguments
- Deleted without affecting copies

````python
## function copy

def welcome():
    return "Welcome to my github profile.."

welcome()
````

````python
wel = welcome
wel()
````

````python
del welcome
````

````python
welcome()
````

````python
wel()
````

## 3. Closure Functions

A **closure** is a nested (inner) function that **remembers the environment** in which it was created — even after the outer function has finished.

This is the foundation of how decorators work.

````python
## closure functions

def main_welcome(msg):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        print(msg)
        print("Please learn these concepts properly")
    return sub_welcome_method()
````

````python
main_welcome("Welcome everyone!")
````

````python
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        func("Welcome everyone to this tutorial")
        print("Please learn these concepts properly")
    return sub_welcome_method()
````

````python
main_welcome(print)
````

````python
def main_welcome(func, lst):

    def sub_welcome_method():
        print("Welcome to python series")
        print(func(lst))
        print("Please learn these concepts properly")
    return sub_welcome_method
````

````python
main_welcome(len, [10, 20, 30, 40, 50])()
````

````python
# Step 1: Define the decorator
def main_welcome(func):

    def sub_welcome_method():
        print("Welcome to the advance python course")
        func()
        print("Please learn these concepts properly")
    return sub_welcome_method
````

````python
# Step 2: Define the function we want to decorate
def course_introduction():
    print("This is an advanced python course")
````

````python
# Step 3: Manually wrap it — this is what the @decorator syntax does under the hood
wrapped = main_welcome(course_introduction)
wrapped()
````

---
## 5. The `@` Syntax

The `@decorator_name` syntax is **syntactic sugar** — it's equivalent to:
```python
course_introduction = main_welcome(course_introduction)
```
Python applies the decorator automatically when the function is defined.

````python
@main_welcome
def course_introduction():
    print("This is the core topics")
````

````python
course_introduction()
````

---
## 4. Building a Decorator Manually

Now let's combine everything: a function that **takes another function**, wraps it with extra behavior, and **returns the wrapped version**.

This is exactly what a decorator is.

````python
# Step 1: Define the decorator

def my_decorator(func):
    def wrapper():
        print("Something is happenign before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
````

````python
@my_decorator
def say_hello():
    print("Hello!!")
````

````python
say_hello()
````

## 6. Decorators with Arguments

To pass **arguments to a decorator**, you need an extra layer of nesting:

```
outer function (takes decorator args)
  └── decorator function (takes the original function)
        └── wrapper function (takes the original function's args)
```

We use `*args` and `**kwargs` in the wrapper so it works with **any function signature**.

````python
## Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
````

````python
@repeat(3)
def say_hello():
    print("Hey!!")
````

````python
say_hello()
````

## 7. Quick Revision Summary

| Concept | Key Takeaway |
|---|---|
| **First-class functions** | Functions are objects; can be assigned, passed, deleted |
| **Closures** | Inner functions capture variables from their outer scope |
| **Decorator** | A function that wraps another function to add behavior |
| **`@` syntax** | Shorthand for `func = decorator(func)` |
| **Decorator with args** | Add an extra outer function layer to pass arguments |
| **`*args, **kwargs`** | Ensures wrapper works with any function signature |

### Mental Model
```
@decorator
def my_func():     ←→     my_func = decorator(my_func)
    ...
```

