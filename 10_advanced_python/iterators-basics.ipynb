# Iterators in Python

Iterators are objects that allow you to traverse through all elements of a collection one at a time, without exposing the underlying structure.


They are useful for:

- Memory-efficient looping
- Working with large datasets
- Lazy evaluation (compute only when needed)

| Concept      | Description                                          |
| ------------ | ---------------------------------------------------- |
| **Iterable** | Object you can loop over (e.g., list, tuple, string) |
| **Iterator** | Object that produces elements one at a time          |


```python
my_lst = [11, 22, 33, 44, 55]
for i in range(len(my_lst)):
    print(my_lst[i])
```

```python
print(type(my_lst))
```

👉 Python automatically creates an iterator internally when using for loops.

```python
## Iterator

iterator = iter(my_lst)
print(type(iterator))
```

```python
iterator
```

```python
## Iterate through all the elements

next(iterator)
```

```python
iterator = iter(my_lst)
```

```python

try:
    print(next(iterator))
except StopIteration:
    print("There are no elements in the iterator.")
```

### Why Iterators?

- Do not store all values in memory
- Generate values on demand
- Useful for large datasets

### Custom Iterator Example

```python
class CountUp:
    def __init__(self, max_value):
        self.current = 1
        self.max = max_value

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.max:
            value = self.current
            self.current += 1
            return value
        else:
            raise StopIteration

counter = CountUp(5)

for num in counter:
    print(num)
```
