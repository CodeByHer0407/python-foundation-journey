## Generators in Python (Memory-Efficient Iteration)

#### Overview
This notebook explains how generators work and why they are useful for handling large datasets efficiently.

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(5):
        yield i**2
```

```python
print(square(3))
```

```python
for i in square(3):
    print(i)
```

```python
a = square(5)
print(a)
```

```python
next(a)
```

```python
def simple_gen():
    yield 1
    yield 2
    yield 3

g = simple_gen()

```

```python
print(next(g))  # 1
print(next(g))  # 2
print(next(g))  # 3
```

```python
for val in g:
    print(val)
```

## ⚙️ How `yield` Works

- `yield` pauses function execution
- Saves state
- Resumes from same point on next call

👉 This makes a generator lazy (on-demand execution)

## Practical Example: Reading Large Files

Generators are paticularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.

```python
### Practical: Reading Large Files

def read_large_file(file_path):
    with open(file_path, "r") as file:
        for line in file:
            yield line
```

```python
file_path = 'large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```

## 🔁 List vs Generator

| Feature | List | Generator |
|--------|------|----------|
| Memory | Stores all values | Generates on demand |
| Speed  | Faster access | Slower per access |
| Use case | Small datasets | Large/stream data |

```python
# List (memory heavy)
nums = [i for i in range(1000000)]

# Generator (memory efficient)
nums_gen = (i for i in range(1000000))
```

```python
import sys
print(sys.getsizeof(nums))
print(sys.getsizeof(nums_gen))
```
